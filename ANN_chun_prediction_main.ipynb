#data prepcocessing

```python
# droping the inrelivent coloum from data set
```

```python
data_set = data_set.drop(["RowNumber","CustomerId","Surname"],axis=1)
```

```python
data_set.head(5)
```

```python
# encoding the categoral coloum from my data_set
```

```python
LB = LabelEncoder()
```

```python
data_set['Gender'] = LB.fit_transform(data_set['Gender'])
```

```python
# 0 = FEMALE
#1 = MALE
```

```python
# so we have one more coloum callled geography , for that we are using an onelot encoder becuse we having 3 catagires .
# if we use labeal encoding it will assign 1,2,3 for three features (france,germoney,spain) but ann things as a number so toslove that we are using onehot
#encoding
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

onehot_encoder_geo = OneHotEncoder()

geo_encoded = onehot_encoder_geo.fit_transform(data_set[['Geography']])

geo_encoded_df = pd.DataFrame(
    geo_encoded.toarray(),
    columns=onehot_encoder_geo.get_feature_names_out(['Geography'])
)

geo_encoded_df
```

```python
data_set = pd.concat([data_set.drop("Geography",axis=1), geo_encoded_df], axis=1)
```

```python
data_set
```

```python

# dividing the data with dependent and independent features
```

```python
data_set.head(5)
```

```python
x = data_set.drop("Exited",axis=1)
y = data_set["Exited"]
```

```python
# splinting the data into train and test
X_train,X_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(type(X_train))
print(type(X_test))

print(X_train.shape)
print(X_test.shape)
```

```python
# scaleing the data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

# ANN  impermentaion

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
```

```python
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
```

```python
# creating a model outline
```

```python
model = Sequential(
    [
        Dense(64,activation='relu',input_shape=(X_train.shape[1],)),
        Dense(32,activation='relu'),
        Dense(1,activation='sigmoid')
    ]
)
```

model.summary()

```python
model.summary()
```

```python

import tensorflow as tf
opt = tf.keras.optimizers.Adam(learning_rate = 0.01)
```

```python
# initilizing the loss and optimizers
```

```python
model.compile(optimizer=opt,loss='binary_crossentropy',metrics=['accuracy'])
```

```python
# initlizing tenserboad
```

```python
import tensorflow as tf
from tensorflow.keras.callbacks import TensorBoard
```

```python
pip install datetime
```

```python
import datetime
from tensorflow.keras.callbacks import TensorBoard

log_dir = "logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")

tensorboard_callback = TensorBoard(
    log_dir=log_dir,
    histogram_freq=1
)
```

```python
# setuping the ealy  stopping technique 
```

```python
early_stopping = EarlyStopping(monitor='val_loss',patience=10,restore_best_weights=True)
```

```python
karthik_model = model.fit(
    X_train,y_train,validation_data=(X_test,y_test),epochs=100,
     callbacks = [tensorboard_callback,early_stopping]
)
```

```python
model.save("model.keras")
```
